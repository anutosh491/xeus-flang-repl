# Inspect the JIT input

`%%mlir` asks the kernel to compile the cell and publish the resulting LLVM-dialect MLIR without executing it. This is the module that the MLIR ExecutionEngine hands to ORC JIT.

```fortran
%%mlir
subroutine saxpy(n, a, x, y)
  integer, intent(in) :: n
  real, intent(in) :: a, x(n)
  real, intent(inout) :: y(n)
  integer :: i
  do i = 1, n
    y(i) = a * x(i) + y(i)
  end do
end subroutine
```
