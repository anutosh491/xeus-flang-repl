# xeus-flang-repl demo

This notebook exercises the native experimental Flang interpreter. Each Jupyter cell is compiled as an incremental MLIR module in one persistent session.

```fortran
print *, "Hello from Flang in Jupyter!"
```

State introduced in one cell remains visible to later cells.

```fortran
integer :: k
k = 1
```

```fortran
do while (k <= 4)
  print *, k
  k = k * 2
end do
```

```fortran
print *, "k after the loop:", k
```

Loose declarations and statements can share a cell; the interpreter wraps them in standard Fortran internally.

```fortran
real :: values(5)
integer :: i
do i = 1, 5
  values(i) = real(i * i)
end do
print *, values
```

A later declaration may shadow an earlier variable without changing normal batch-mode Flang semantics.

```fortran
integer :: k
k = 100
print *, "redefined k:", k
```
